### Diffusion equation and it's solution

If we define a constant diffusion parameter $D$ as

$$
D = \frac{dh^2}{2 \, \Delta t},
$$

and then take the limit $\Delta t, h \to 0$, we obtain the diffusion equation for constant diffusion:

$$
\frac{\partial \phi(x,t)}{\partial t} = D \frac{\partial^2 \phi(x,t)}{\partial x^2}.
$$

### 1a

Show that the following function is a solution of the diffusion equation:

$$
\phi(x,t) = \frac{1}{\sqrt{4 \pi D t}} \int_{-\infty}^{\infty} h(y) \, e^{-\frac{(x-y)^2}{4Dt}} \, dy.
$$

This is a convolution integral and can be analyzed using Fourier transforms:

$$
(f * g)(x) = \int_{-\infty}^{\infty} f(y) \, g(x-y) \, dy = \int_{-\infty}^{\infty} f(x-y) \, g(y) \, dy,
$$

together with the rule

$$
\mathcal{F}\{ f * g \}(\omega) = \sqrt{2 \pi} \, \hat{f}(\omega) \, \hat{g}(\omega).
$$

We can rewrite $\phi(x,t)$ as a convolution:

$$
\phi(x,t) = \frac{1}{\sqrt{4 \pi D t}} \, (h * e^{-x^2/(4Dt)})(x).
$$

---

Fourier Transform

Let

$$
\mathcal{F}\{h(x)\} = \hat{h}(\omega),
$$

and use the standard Fourier transform of a Gaussian:

$$
\mathcal{F}\{ e^{-a x^2} \} = \frac{1}{\sqrt{2a}} \, e^{-\frac{\omega^2}{4a}}.
$$

Then

$$
\mathcal{F}\{\phi(x,t)\} = \mathcal{F}\left\{ \frac{1}{\sqrt{4 \pi D t}} \, h * e^{-x^2/(4Dt)} \right\}(\omega) 
= \frac{\sqrt{2 \pi}}{\sqrt{4 \pi D t}} \, \hat{h}(\omega) \cdot \frac{1}{\sqrt{2a}} \, e^{-\frac{\omega^2}{4a}}.
$$

Substituting $a = \frac{1}{4Dt}$ gives

$$
\mathcal{F}\{\phi(\omega,t)\} = \hat{\phi}(\omega,t) = \hat{h}(\omega) \, e^{-D t \omega^2}.
$$

---

Verification

Fourier transforming the diffusion equation:

$$
\mathcal{F}\Big\{ \frac{\partial \phi(x,t)}{\partial t} \Big\} = \frac{\partial \hat{\phi}(\omega,t)}{\partial t}, 
\quad
\mathcal{F}\Big\{ D \frac{\partial^2 \phi(x,t)}{\partial x^2} \Big\} = -D \omega^2 \, \hat{\phi}(\omega,t).
$$

Differentiating $\hat{\phi}(\omega,t)$ with respect to $t$:

$$
\frac{\partial}{\partial t} \hat{\phi}(\omega,t) = \frac{\partial}{\partial t} \left( \hat{h}(\omega) \, e^{-D t \omega^2} \right) = - D \omega^2 \, \hat{h}(\omega) \, e^{-D t \omega^2}.
$$

Thus, both sides of the Fourier-transformed diffusion equation are equal, which implies that the original equation is satisfied for $(\phi(x,t))$. This can also be verified by taking the inverse Fourier transform.

---
Physical interprotation of h(y)

Letting $t \to 0$ in our fouriertransformed function $\hat\phi(\omega,t) = \hat h(\omega)\cdot e^{-Dt\omega^2}$ we get the following initial condition in the $\omega$-plane:

$$ \lim_{t \to 0} \hat\phi(\omega,t) = \hat\phi(\omega,0) = \lim_{t \to 0} \hat h(\omega)\cdot e^{-Dt\omega^2} = \hat h(\omega)$$

Applying the inverse Fouriertransform: 

$$\phi(x,0) = \mathcal{F}^{-1}{[\hat{\phi}(\omega,0)](x)} = \mathcal{F}^{-1}{[\hat h(\omega)](x)} = h(x) $$

Which yields our initial condition when t = 0: 

$$\phi(x,0) = h(x)$$

Meaning that $h(x)$ describes the initial particle distribution in our system. 

We can also show that this is the case by taking the limit $t \to 0$ in

$$ \lim_{t \to 0} \phi(x,t) = \lim_{t \to 0} \int_{-\infty}^{\infty} h(y)\cdot e^{\frac{-(x-y)^2}{4Dt}} dy= \int_{-\infty}^{\infty} h(y)\cdot \lim_{t \to 0}e^{\frac{-(x-y)^2}{4Dt}} dy$$

As they explain in the project description under "Analytical solution" the dirac delta function $\delta(x)$ as our initial particle distribution yields a normalised solution to our system on the form
$$\phi(x,t) = \frac{1}{\sqrt{4\pi Dt}}e^{\frac{-(x-y)^2}{4Dt}}, t>0.$$

This means that our limit $t \to 0$ yields $\lim_{t \to 0} \phi(x,t) = \phi(x,0) = \delta(x)$. 
Then 

$$
\lim_{t \to 0} \phi(x,t) = \int_{-\infty}^{\infty} h(y)\cdot \lim_{t \to 0}e^{\frac{-(x-y)^2}{4Dt}} dy= \int_{-\infty}^{\infty} h(y)\cdot \delta(x-y) dy = h(x)
$$ 
Which again gives the inital condition $$\phi(x,0) = h(x)$$ to our system. 

---
Conclusion 

$\phi(x,t)$ given by

$$
\phi(x,t) = \frac{1}{\sqrt{4 \pi D t}} \int_{-\infty}^{\infty} h(y) \, e^{\frac{-(x-y)^2}{4Dt}} \, dy
$$

is a solution to the diffusion equation for any initial distribution $h(x)$, as shown by applying the Fouriertransform.

### 1b


If the probability of finding a certain particle at a position x in a potential landscape is given by: 

$$ 
P(x) = e^{\frac{-\beta\cdot V(x)}{Z}}
$$

where $V(x)$ is the potential at a given point, Z the canonical partition function, and  $\beta = \frac{1}{k_b\cdot T}$ 

Then we can define the probabilities for where a particle may end up after a time step by finding the probabilities of each position and dividing by the total probabilities for all positions. 

Since we have defined that the particle only can move either one step to the right, one to the left, or stay put, there are in total three legal positions for where the particle may end up. 

This means that the probability of the particle going to the left, hereby denoted by $p^{-}$, can be described by: 

$$p^{-} = \frac{P(x-1)}{P(x-1)+P(x)+P(x+1)}$$

Filling in for each of the probabilities we get: 

$$p^{-} = \frac{e^{\frac{-\beta\cdot V(x-1)}{Z}}}{e^{\frac{-\beta\cdot V(x-1)}{Z}} + e^{\frac{-\beta\cdot V(x)}{Z}} + e^{\frac{-\beta\cdot V(x+1)}{Z}}}$$

If the probability of a particle moving to a position $x$ from a position $x_0 \in ({x-1,x,x+1})$ is proportional to $P(x)$ and we divide by $e^{\frac{-\beta\cdot V(x-1)}{Z}}$ in both numerator and denominator we get the final expression: 

$$p^{-} = \frac{1}{1+e^{-\beta\cdot[V(x_0)-V(x_0-1)]} + e^{-\beta\cdot [V(x_0+1)-V(x_0-1)]}}$$

Using the exact same procedure one obtains the formula for both $p^+$ and $p^0$ $$p^+ = \frac{1}{1+e^{-\beta\cdot[V(x_0)-V(x_0+1)]} + e^{-\beta\cdot [V(x_0-1)-V(x_0+1)]}}$$ $$p^0 = \frac{1}{1+e^{-\beta\cdot[V(x_0+1)-V(x_0)]} + e^{-\beta\cdot [V(x_0-1)-V(x_0)]}}$$

### 1c


The first condition $k_bT >> |{V(x+1)-V(x)}|$ for all x:

This condition physically describes a system where the particles have much energy and the potential differences are very small compared to that energy. This makes the system act like a system with a constant potential and the probabilities become $$p^- = p^+ = p^0 = \frac{1}{3}$$

By using the condition we can show that $e^{-\beta\cdot[V(x_0+1)-V(x_0)]} = e^{\frac{-[V(x_0+1)-V(x_0)]}{k_bT}}$ equals approximatly 1. For all the exponentialts the exponential goes to 0 when $k_bT >> |{V(x+1)-V(x)}|$, this is easy to see when the potentials lie one step apart. Using the triangle inequality we can show that the exponentials go to 0 when the potential difference lie to steps apart: 

$$|V(x_0-1)-V(x_0+1)| = |V(x_0+1)-V(x_0)+V(x_0)-V(x_0-1)| \leq |V(x_0+1)-V(x_0)| + |V(x_0)-V(x_0-1)| << k_bT + k_bT = 2k_bT$$

Then we can rewrite the exponential $\frac{V(x_0+1)-V(x_0-1)}{k_bT}$ as $\frac{2\cdot(V(x_0+1)-V(x_0-1))}{2k_bT}$ which equals $2\cdot0$ in the limit $k_bT >> |{V(x+1)-V(x)}|$

The second condition, $k_bT << |{V(x+1)-V(x)}|$ for all x, describes a system where the potential differences are way bigger than the thermic energy in the system, resulting in a system where all the particles get trapped in local potential wells. 

$$p^i = \frac{e^{-\beta\cdot V(x_i)}}{e^{-\beta\cdot V(x_0-1)} + e^{-\beta\cdot V(x_0)} + e^{-\beta\cdot V(x_0+1)}}$$

Where $x_i \in (x_0-1,x_0,x_0+1)$

The denominator can be simplified in the case where one of the potentials is lower than the others. We can then write $$\frac{e^{-\beta\cdot V_{higher}}}{e^{-\beta\cdot V_{higher}} + e^{-\beta\cdot V_{lower}} + e^{-\beta\cdot V_{higher}}}$$

Each of the $e^{-\beta V_{higher}}$ expressions can be rewritten to $e^{-\beta V_{lower}}\cdot e^{-\beta(V_{higher}-V_{lower})}$ and since the difference between the potentials is much bigger than the thermal energy in the system, $e^{-\beta(V_{higher}-V_{lower})} = e^{-\frac{V_{higher}-V_{lower}}{k_bT}} = e^{-\infty} = 0$ Therefore all the $V_{higher}$ expressions go to 0 and only the $V_{lower}$ expressions remain. 

$$
p^i = \frac{0}{e^{-\beta V_{lower}}} = 0 \lor p^i = \frac{e^{-\beta V_{lower}}}{e^{-\beta V_{lower}}} = 1
$$ 

So: 

$$
p^i \to
\begin{cases}
1, & \text{if } V(x_i)=V_{\min}, \\
0, & \text{otherwise}.
\end{cases}
$$

The particles will with 100% probability move towards the point with the lowest potential of the three possible points. Over time this will result in all the particles getting stuck in the local potential minima, or the global minima if the potential is linear.

```python
import numpy as np
import matplotlib.pyplot as plt

h = 1
delta_t = 1
num_particles = 101
num_time_steps = 200
num_simulations = 100
D = h**2/(3*delta_t)

def change_in_pos(p_minus, p_plus):
    p = np.random.uniform(0,1)
    if p < p_plus:
        delta_pos = 1
    elif p > (1 - p_minus):
        delta_pos = -1
    else:
        delta_pos = 0
    return delta_pos


def calculate_end_pos_no_repulsion(p_minus, p_plus):
    x_pos = np.arange(-50, 50 + 1, h) # Initial positions
    for _ in range(num_time_steps):
        
        # Creates a random number on [0, 1) for every particle.
        random_array = np.random.uniform(0, 1, num_particles)
        # Creates an array of True/False.
        move_minus = random_array < p_minus
        move_plus = random_array > 1 - p_plus
        # Changes position where True occures.
        x_pos[move_minus] -= h
        x_pos[move_plus] += h
    return x_pos
    

def calculate_end_pos_hard_core_repulsion(p_minus, p_plus):
    rng = np.random.default_rng()
    particles = np.arange(1, 101 + 1)

    cytoskeleton = np.array([])
    cytoskeleton = np.append(cytoskeleton, particles)

    # print(cytoskeleton)

    for _ in range(num_time_steps):
        # Go trough the particles in random order.
        rng.shuffle(particles)
        for particle in particles:
            
            # Calculate the particles position change attempt.
            delta_pos = change_in_pos(p_minus, p_plus)
            
            # If the particle don't try to move. Continue to next particle.
            if delta_pos == 0:
                continue
            
            index_particle = np.where(cytoskeleton == particle)[0][0]
            
            # If particle tries to move outside (on the left) of the cytoskeleton. Expand the skeleton on both sides.
            # Expanding on both sides keeps the initial center of the cytoskeleton at the center after every expantion.
            if index_particle + delta_pos == -1:
                cytoskeleton = np.insert(cytoskeleton, 0, 0)
                cytoskeleton = np.append(cytoskeleton, 0)
                
                # New particle index
                index_particle = np.where(cytoskeleton == particle)[0][0]
            
            while True:
                # This try except block makes sure that we don't have to know the cytoskeleton-array's final size before each simulation.
                try:
                    #Check if move is possible. Move if possible.
                    if cytoskeleton[index_particle + delta_pos] == 0:
                        cytoskeleton[index_particle + delta_pos] = cytoskeleton[index_particle]
                        cytoskeleton[index_particle] = 0
                    break
                except IndexError:
                    # Expand the skeleton on both sides.
                    cytoskeleton = np.insert(cytoskeleton, 0, 0)
                    cytoskeleton = np.append(cytoskeleton, 0)
                    index_particle += 1 # Adding 1 to index because of the extra zero added to cytoskeleton.
                
    # print(cytoskeleton)

    end_positions = np.array([])
    center_index = (len(cytoskeleton) - 1) // 2
    for i in range(len(cytoskeleton)):
        if cytoskeleton[i] != 0:
            end_positions = np.append(end_positions, (i - center_index)*h)

    return end_positions
```

```python
p = 1/3

# Simulate num_simulations times
total_end_pos = np.array([])
for s in range(num_simulations):
    total_end_pos = np.append(total_end_pos, calculate_end_pos_no_repulsion(p, p))

# Plotting

bins = np.arange(total_end_pos.min(), total_end_pos.max() + 2)
x_pos2 = np.linspace(total_end_pos.min(), total_end_pos.max(), 400)
weights = np.ones_like(total_end_pos) / num_simulations


def Analytical(x):
    result = np.zeros_like(x, dtype=float)
    
    for x0 in range(-50, 51):
        result += np.exp(-(x-x0)**2/(4*D*200))
        
    result *= 1/(101*np.sqrt(4*np.pi*D*200))
    
    return result
plt.hist(total_end_pos, bins, weights=weights)
plt.title("Plot 1: No repulsion, V(x) = k, B*k = 1")
plt.xlabel("X-position")
plt.ylabel("Average occurrence of the particles")
plt.show()

plt.hist(total_end_pos, bins, density = True, weights=weights, label = "SPD")
plt.plot(x_pos2, Analytical(x_pos2), color = "red", linestyle = "-", label = "AAGD")
plt.title("Plot 2: Simulated vs. Analytical")
plt.xlabel("X-position")
plt.ylabel("Probability P(X = x)")
plt.legend()
plt.show()
```

Plot 1 and 2 describes a system with No repulsion, V(x) = k, B*k = 1.
The first plot shows the average position for every particle after 100 simulations, which is found by adding the final positions after each simulation, and dividing by total amount of simulations. 
The second plot shows a normalized ($\int_{-\infty}^{\infty} \phi(x,t) dx = 1$) distribution of the average particle positions after 100 timesteps (SPD, Simulated Particle Distribution). Additionaly it includes the analytical distribution for the simulation which is the average of each particles gaussian distribution (AAGD, Average Analytical Gaussian Distribution). A single particle starting in the position $x_0$ will follow a normal distribution $\phi(x,t) = \frac{1}{\sqrt{2\pi}\cdot \sigma} exp(\frac{\mu^2}{2\sigma^2})$ with $\mu = (x-x_0), \sigma = \sqrt{2Dt}$. As we can see the average excpected position for each particle is its starting position in a flat potential, and the standard deviation is proportional to the amount of timesteps we take, $t$. To find the total distribution for all the particles we find the average of all the particle distributions. $\phi(x,t) = \frac{1}{101} \sum_{x_0 = -50}^{50} \frac{1}{\sqrt{4\pi Dt}}e^{\frac{-(x-x_0)^2}{4Dt}}$.

### Comparison of no repulsion and hard core repulsion:
The particle distribution in a)i) (constant potential, no repulsion) and b)i) (constant potential, hard core) have roughly similar shapes. We see that the extreme points are quite similar, where some particles have reached positions as far as $\pm75$ in both cases. The fact that these "extreme cases" are not significantly affected by the presence/absence of repulsion between the particles makes sense, because it is the particles at the far left and far right (with hard core repulsion) that move most freely. And therefore also most similarly to the particles with no repulsion. $\newline$ In the region of the initial positions we see larger differences. Here the no repulsion distribution is characterized by randomness arising from the fact that the particles move independently of each other, with $p^{-} = p^{0} = p^{+} = \frac{1}3$. With hard core repulsion the distribution is much more even and actually completely flat near the center. Not enough time has passed for the particles in the middle to have space to move.

It is under the potential $V(x) = -kx$ that the differences become clear. The particle distributions in a)ii) and b)ii) are completely different. The linear potential gives a significantly larger probability that particles will move to the right. The problem for the particles with hard core repulsion (b ii) is that they block each other. We see that the particles furthest to the right reach approximately the same region as the rightmost edge of the particles in a)ii). Apart from that, we see that all the other particles (those not at the far right) must constantly wait until there is space to move. The leftmost edge has barely moved because they mostly have to wait.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import erfc

bk = 1000

# Time
t_end = 10_100
T_p = 500   # Timesteps for V1-V2 switch

# Simulation
h = 1 #step length
alpha = 0.8

amount_of_N_x = 2
N_x = 100   # Period of V(x)
N_p = 12*N_x

n_minus = np.zeros(shape=(t_end), dtype=int)
n_plus = np.zeros(shape=(t_end), dtype=int)


# Simulation functions

def p_plus(x, V):
    return 1/(1+np.exp(-bk*(V(x-1)-V(x+1)))+np.exp(-bk*(V(x)-V(x+1))))

def p_minus(x, V):
    return 1/(1+np.exp(-bk*(V(x+1)-V(x-1)))+np.exp(-bk*(V(x)-V(x-1))))

def sawtooth_potential(x: np.array):
        rest_x = np.mod(x,N_x)

        return np.where((rest_x > 0) & (rest_x <= alpha * N_x),
                        rest_x / (N_x * alpha),
                        -(rest_x-N_x) / ((1 - alpha) * N_x))

def constant_potential(_):
    return 1

```

```python

def generate_start_pos(amount_N_x,N_p, mode="3B"):
    # Generates start posistions of the particles. The value of index i corresponds to particle it's x-posisiton.
    
    startpos_array = np.zeros(shape=(N_p), dtype=int)
    
    if mode == "3A": 
    # 3A
    # Particles placed evenly on x-axis
        for i in range(N_p):
            startpos_array[i] = i // int(N_p/N_x/amount_N_x)
    else:
    # 3B
    # Minima of sawtooth is given by 0 and N_x
        startpos_array[:int(startpos_array.size/2)] = 0
        startpos_array[int(startpos_array.size/2):] = N_x 


    
    return startpos_array


def randomwalk_step(x: np.array, V, index):
    r = np.random.rand(x.size)
    p_m = p_minus(x, V)
    p_p = p_plus(x, V)

    x_new = x.copy()

    move_minus = r <= p_m
    move_plus  = r >= (1 - p_p)
    # Compares elementwise^

    n_plus[index] = move_plus.sum()
    n_minus[index] = move_minus.sum()

    x_new[move_minus] -= h
    x_new[move_plus]  += h

    x_new = np.mod(x_new, amount_of_N_x * N_x)

    return x_new


def random_walk_ratchet_potentital(t_end: int, T_p: int, amount_N_x: int, mode="3B") -> np.array:
    positions = generate_start_pos(amount_N_x, N_p, mode=mode)
    
    for t in range(t_end):
        if (t // T_p) % 2 == 0:
            # Constant pot, starts with this one
            positions = randomwalk_step(positions,constant_potential, index=t)
        else: 
            # Sawtooth pot
            positions = randomwalk_step(positions,sawtooth_potential, index=t)
    
    return positions
```

```python
# Redefining variables to make sure that the block simulates correctly

# Physics
bk = 1000

# Time
T_p = 500   # Timesteps for V1-V2 switch
t_end = 20*T_p

# Simulation
h = 1 #step length
alpha = 0.8

amount_of_N_x = 2
N_x = 100   # Period of V(x)
N_p = 12*N_x

n_minus = np.zeros(shape=(t_end), dtype=int)
n_plus = np.zeros(shape=(t_end), dtype=int)


final_pos = random_walk_ratchet_potentital(t_end, T_p, amount_of_N_x, mode="3A")


# Plotting
fig, ax1 = plt.subplots()

# Histogram on the left
ax1.hist(final_pos, bins=100, label="Number of particles")
ax1.set_ylabel("Number of particles")
ax1.legend(loc="upper left")

# Potential on the right
ax2 = ax1.twinx()
x_array = np.linspace(1, N_x*amount_of_N_x - 1, 1000)
ax2.plot(x_array[:-1], sawtooth_potential(x_array[:-1]), linewidth=2, color="orange", label="Potential")
ax2.set_ylabel("Potential")
ax2.legend(loc="upper right")

ax1.set_xlabel("Positions")
plt.title(rf"Final positions for 10000 steps with $\alpha = 0.8$")


j = n_plus - n_minus
j_avg = []
cycle_length = int(t_end/10) 


print("Values for J_avg for each cycle:")
for i in range(10):
    j_avg.append(float(j[cycle_length*i: cycle_length*(i+1)].sum()/cycle_length/N_p))
    print(f"j_avg({i}) = {j_avg[-1]}")

```

As we can see the particles behave as expected and end up in the bottom of the potentials. The reason we get a non-zero current is that while the particles are in a constant potential, they spread around evenly and become wider and wider, and as soon as the potential switches the particles (on average) end up on a slope that is pushes it to the left (negative direction of course). In other words, the particles more often than not end up on the other side of the steep potential hill. This causes a negative current.

We can also see that this current is larger for the first cycle. This is because the particles are initially spread out evenly. When the potential changes for the first time to a sawtooth, most of the particles (on average alpha=80% of the particles) land on a leftwards slope, this causes a larger current for the first cycle. Wheras for all other cycles, the particles are more likeley to stay in the bottom of the pit.

For this first graph, the only real thing we can see is that for a very short interval (=1) the current becomes large and positive and for larger intervals we get a smaller negative current. 

For the case of 2 time steps we can go through it manually. All the particles start at the bottom of the 2 potential pits. The first time step is a constant potential, so the particles evenly split between moving (-1,0,1). For the second step, the particles are more likely to move down the steeper hill, which causes the particles to move (on average) a lot per cycle. For all the other cases, they can be graphed seperately to see them better:

```python
plt.plot(T_p_array[1:], j_avg[1:])
plt.xlabel("Voltage interval T_p")
plt.ylabel("Average particle stream")
plt.grid()
plt.title("Average particle stream vs. voltage period")
plt.show()
```

Now we can see the graph more clearly. The reason the simulation forms this curve is that for small values of T_p, the particles don't have enough time during the constant potential to spread "over the steep hill" to reach the other side, so they (on average) stay put in one of the wells. For the smallest value of T_p, the particles almost never make it over the steep hill, and the larger probability of falling down the steep hill takes over, giving a positive current.

The larger the value of T_p you get, the more likely the particles are to go over the hill, until you reach an optimal value of around 500 (which, coincidentally, was the starting value). After this peak, the current slows down. This slowdown is caused by the fact that the particles spend more time at the bottom of the potential well, which causes the particles to move less on average.

After the optimal value at around 500, the average particle stream will go asymptotically towards 0. This is due to the particles being stuck in the potential wells for longer, being unable to move, and we divide by t_end = 2*T_p, which results in the current being antiproportional to T_p.

```python
# Physics
bk = 1000

# Time
T_p = 500 
t_end = 2*T_p

# Simulation
h = 1 #step length
alpha_array = np.linspace(0.001,0.999,50)

amount_of_N_x = 2
N_x = 100   # Period of V(x)
N_p = 12*N_x
alpha = 0.8


n_minus = np.zeros(shape=(t_end), dtype=int)
n_plus = np.zeros(shape=(t_end), dtype=int)
j_avg = []
```

Analytical solution

```python

def J_avg_analytical(N_x, T_p, alpha):
    return (N_x / (4 * T_p)) * (erfc((alpha * N_x) / 2 * np.sqrt(3/T_p)) - erfc(((1 - alpha) * N_x) / 2 * np.sqrt(3/T_p)))

J_analytical_values = J_avg_analytical(N_x, T_p, alpha_array)


plt.plot(alpha_array, J_analytical_values, label = "Analytical")
plt.title("Analytical J_avg as func of alpha")
plt.xlabel("alpha")
plt.ylabel("J_avg")
plt.legend()
plt.grid()
plt.show()
```

Numerical solution

```python
bk = 1000
n_minus = np.zeros(shape=(t_end), dtype=int)
n_plus = np.zeros(shape=(t_end), dtype=int)
j_avg = []

for alpha_ in alpha_array:
    alpha = alpha_

    n_minus = np.zeros(shape=(t_end), dtype=int)
    n_plus = np.zeros(shape=(t_end), dtype=int)

    random_walk_ratchet_potentital(t_end, T_p, amount_of_N_x)

    j = n_plus - n_minus
    j_avg.append(j.sum()/(t_end*N_p))
    
plt.plot(alpha_array, j_avg, label="Numerical")
plt.title("Numerical J_avg as func of alpha")
plt.xlabel("alpha")
plt.ylabel("J_avg")
plt.legend()
plt.grid()
plt.show()
```

The first thing we observe is that the numerical and analytical calculation have almost the exact same values. This means that our numerical program fits the analytical model. 

The reason we get these results is that for values of alpha close to 1 or 0 the particles are very likely to land on the top of the slopes, which results in a larger average current in either the positive or negative direction. We also see a zero current for alpha = 0.5, as this gives just as likely movement in each direction.

We decided to split the values of $\beta k$ in two graphs to get a better look at the various effects we can see.

$\beta k$ refers to the strength of the potential. For large values you would expect the particles to always move downhill (like the analytical solution assumes), however for smaller values the particles also have a higher chance to move upwards, and this we can see in this graph. This is shown best in $\beta k = [10, 0.01]$, where the first case looks a lot like the analytical solution, only scaled down by a constant factor. This can be expected as the particles move less down the hill. For the second case of $\beta k = 0.01$ we can see that the strength of the potential is so low that the particles essentially move around randomly, and the different values of alpha change (essentially) nothing of the particle current.

The more interesting results are the ones we can see for the medium strengths (1,2,3). We see some similarities for the values, they are positive and negative for the extreme values of alpha, this makes sense as the particles don't have any potential pushing them in the other direction. They also have a current of 0 for $\alpha = 0.5$, which obviously makes sense. 

The interesting part are the values between, where we get a small current in the opposite direction from what we get for larger $\beta k$. The reason for this is that the probability of a particle moving in a certain direction is given by the difference in potentials to a particles right and left. And for these small value of $\beta k$ the voltage difference for the non-steep slope for these medium values of $\alpha$ is negligible, and the particles only get pushed by the larger potential differeces given by the steep slopes of the sawtooth potential. 

We also see how this effect gets weaker for weaker strengths, as this effect is slightly weaker for $\beta k = 1$ and virtually non-existant for $\beta k = 0.01$.

### 3F

I think the biggest flaw in the logic presented in the analytical solution is that it doesn't take into account the statistics.
The logic that is presented works well for a ball rolling down the "hill" that the potential is, but it doesn't work for out particles that have a low, but crucially real, chance of going "up the hill"
If there was a 100% chance that the particles would move in the correct direction, the argument would hold, and you would get the exact same solution.

However, the more likely the particles are to move to a higher potential, the less analytical our simulation becomes. 
This is what we see in 3d, where the particles generally move in the correct direction (correct derivatives in start an end), but generally are far from the analytical solution

This only goes up to a point however, as even if we set bk to essentially infinity, we still get fluctiations when the particles aren't at the bottom of the potential well.
This can be seen if you set bk to an extreme value like 10**20.

```python
# Physics
bk = 1000

# Time
t_end = 200
T_p = 40   # Timesteps for V1-V2 switch

# Simulation
h = 1 #step length
alpha = 0.2

amount_N_x = 5
N_x = 20   # Period of V(x)
N_p = 10

b = 5

# 4B Variables
N_s = 5
bk = 1000
N_x = 100
T_p = 300
alpha = 0.2
b = 20
N_c = 100 # Number of cycles
rho = b*N_p / (N_s*N_x)


n_plus, n_minus = np.zeros(N_c*T_p), np.zeros(N_c*T_p)
x_axis_len = N_s*N_x
```

```python

def p_plus(x, V):
    return 1/(1+np.exp(-bk*(V(x-1)-V(x+1)))+np.exp(-bk*(V(x)-V(x+1))))

def p_minus(x, V):
    return 1/(1+np.exp(-bk*(V(x+1)-V(x-1)))+np.exp(-bk*(V(x)-V(x-1))))

def sawtooth_potential(x: np.array):
        rest_x = np.mod(x,N_x)

        return np.where((rest_x > 0) & (rest_x <= alpha * N_x),
                        rest_x / (N_x * alpha),
                        -rest_x / ((1 - alpha) * N_x) + 1/(1-alpha))

def constant_potential(_):
    return 1

def generate_start_pos_interactions(N_p: int, x_axis_len, b):
    # Returns an array with length N_p evenly spaced. Raise an error if N_p too large to fit b-wide particles
    if N_p * b > x_axis_len:
        raise "Not enough room to generate particles"
    else:
        return np.linspace(0,x_axis_len, num=N_p, dtype=int)


def randomwalk_step_interaction(x: np.array, V, index, x_axis_len, b):
    particle_order = np.arange(x.size)
    np.random.shuffle(particle_order)

    p_m = p_minus(x, V)
    p_p = p_plus(x, V)

    r = np.random.rand(x.size)
    move_minus = r <= p_m
    move_plus  = r >= (1 - p_p)
    # Compares all elements^

    for i in particle_order:
        if move_minus[i] and (x[i] - x[i-1]) % (x_axis_len) > b:
            # Move backwards and not too close
            x[i] = np.mod(x[i] - h, x_axis_len)
            n_minus[index] += 1

        elif move_plus[i] and (x[(i+1)%N_p] - x[i]) % (x_axis_len) > b:
            # Move forwards and not too close
            x[i] = np.mod(x[i] + h, x_axis_len)
            n_plus[index] += 1

    return x

def random_walk_ratchet_potentital_interaction(t_end: int, T_p: int, amount_N_x: int, b: int) -> np.array:
    positions = generate_start_pos_interactions(N_p, amount_N_x*N_x, b)
    x_axis_len = amount_N_x * N_x
    pos_history = np.zeros(shape=(N_p, t_end))

    for t in range(t_end):
        if (t // T_p) % 2 == 0:
            # Constant potential, starts with this one
            positions = randomwalk_step_interaction(positions, constant_potential, t, x_axis_len, b)
        else: 
            # Sawtooth potential
            positions = randomwalk_step_interaction(positions, sawtooth_potential, t, x_axis_len, b)

        pos_history[:, t] = positions
    
    return positions, pos_history
```

## 4A

```python
bk = 1000 # Strength of potential, the only one that's used

# Time
T_p = 40   # Timesteps for V1-V2 switch
t_end = 5*2*T_p

# Simulation
h = 1 #step length
alpha = 0.2

amount_N_x = 4
N_x = 20   # Period of V(x)
N_p = 10

b = 5 # Width of particles

final_pos, pos_history = random_walk_ratchet_potentital_interaction(t_end, T_p, amount_N_x, b)

for i in range(N_p):
    plt.plot(pos_history[i,:], np.arange(t_end), marker=".", linestyle='None')

plt.grid()
plt.title("Position of hard core particles moving in ratchet potential")
plt.xlabel("Position on x-axis")
plt.ylabel("Timesteps")
plt.plot()
```

We can clearly see the location in which the potential becomes a sawtooth.
The particles reach an equilibrium where they can't move down the well because of another particle is in the way, and are unlikely to move up the well.
And as soon as the sawtooth potential is over, the particles begin to move around, with fewer collisions.

## 4D

The paper we decided to compare our simulation to is the paper "Interacting particles in disordered flashing ratchets" by Chacko and Tripathy (Chacko, J., Tripathy, G. Interacting particles in disordered flashing ratchets. Indian J Phys 89, 981–988 (2015). https://doi.org/10.1007/s12648-015-0660-5). And more specifically compare our results to Fig. 2 (a):

Figure text says the following: (a) Plot of $J$ versus $\rho$ for the discrete flashing ratchet (FR) for $a=1$ (filled circles) and $2$ (open circles) for $\omega = 0.05$. The plots are approximately of the form $J(\rho) = A_0\rho(1-\rho)$. 

The nice thing about this paper is that the first parts are very understandable and similar to our project. The authors set up particles in a 1D simulation with a certain width and change the potential between a constant (where the particle randomly diffuse outwards) and a sawtooth potential. The only difference is that is seems they only use relative values, so it's not intuitive how we could easily get the exact same values for our simulation.

We have already almost made fig 2(a) already, the only difference is that we don't have the correct value of T_p, which we have now found out changes the value of where $j_{avg}$ peaks. However, we found out that by choosing a T_p of around 200 we get and extremely good match. We plotted it below together with the function the autors gave in the figure text for $J(\rho)$. We also cheat and find the value of $A_0$ based on the graph, as the authors say nothing about how to get this value.


```python
N_s = 10
bk = 1000
N_x = 100
T_p = 200
alpha = 0.2
b = 20

x_axis_len = N_s*N_x
N_p_array = np.linspace(0, N_s*N_x/b, num=10, dtype=int)
j_avg = []

N_c = int(50_000/T_p)

for N_p_ in N_p_array:
    N_p = int(N_p_)
    n_plus, n_minus = np.zeros(N_c*T_p), np.zeros(N_c*T_p)

    _,_ = random_walk_ratchet_potentital_interaction(N_c*T_p, T_p, N_s, b)
    
    j = n_plus - n_minus
    j_avg.append(j.sum()/(N_p*N_c*T_p + 1e-6)) # Add a small value to not divide by 0, this value is virtually nothing for all values


plt.xlabel(rf"Particle denisty $\rho$")
plt.ylabel("J_avg")
plt.title("Particle stream as a func of number of particles")
plt.grid()

A_0 =  4 * np.max(j_avg)
rho = np.linspace(0,1)
plt.plot(rho, A_0*rho*(1-rho), label="Analytical")

plt.plot(N_p_array * b / (N_s*N_x), j_avg, label="Numerical")
plt.legend()
plt.show()
```

We get a numerical simulation that closely matches the analytical value the authors get. Our simulation also has a larger variation for smaller values, however it fits extremely well for the larger denisities. We also get a maximum which is essentially the same as the authors.

Based on all of this data it would be safe to say that our simulation can recreate the results of other papers. Even though we had to manually choose $A_0$.
